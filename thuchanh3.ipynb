```python
n = int(input())
if 100 <= n < 1000:
  a = n // 100
  b = n // 10 - a * 10
  c = n - b * 10 - a * 100
  dao = c * 100 + b * 10 + a
  print(dao)
else:
  print("k ton tai")
```
