```python
a = str(input())
# Hàm ord() chỉ mong đợi một ký tự duy nhất.
# Nếu người dùng nhập nhiều hơn một ký tự, chúng ta sẽ lấy ký tự đầu tiên
# để tránh lỗi TypeError.
if len(a) > 0: # Đảm bảo chuỗi không rỗng
    b = ord(a[0]) # Lấy ký tự đầu tiên
    c = b - 32
    d = chr(c)
    print(d)
else:
    print("Không có đầu vào nào được cung cấp.") # Xử lý trường hợp đầu vào rỗng
```
