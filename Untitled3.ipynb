```python
I = map(int,input().split())
rs = []
for i in I:
  if i not in rs:
    rs.append(i)
print(rs)
```

```python
n = map(int,input().split())
tong = 0
rs = []
for i in n:
  tong += i
  rs.append(tong)
print(rs)
```

```python
tup = tuple(map(int, input().split()))
k = int(input())
n = len(tup)
k = k % n
kq = tup[k:] + tup[:k]
print(kq)
```
