```python
def giaithua(x):
  s = 1
  for i in range(2,x+1):
   s *= i
  return s
s = giaithua(5)
print(s)
```

```python
def giaithua(x):
  s = 1
  for i in range(2,x+1):
   s *= i
  return s

t = 0
n = int(input('n = '))
for i in range(1,n+1):
  t += giaithua(i)
print(t)
```
